# Concept:- 
* ###  df.groupby('column')['value_column'].aggregation()

```python
import pandas as pd
df = pd.read_csv('titanic.csv')
```

```python
# Q1: Har gender mein average age kya thi?

df.groupby('Sex')['Age'].mean()
```

```python
# Q2: Har Pclass mein survival rate kya thi? (mean of Survived)

df.groupby('Pclass')['Survived'].mean()
```

```python
# Q3: Har gender mein kitne log survive kiye? (sum of Survived)

df.groupby('Sex')['Survived'].sum()
```

```python
# Q4: Har Pclass mein average fare kya thi — descending order mein dikhao?

df.groupby('Pclass')['Fare'].mean().sort_values(ascending=False)
```

```python
# Q5: Har Pclass aur Gender combination mein average survival rate kya thi? (groupby mein do columns)

df.groupby(['Pclass', 'Sex'])['Survived'].mean()
```

```python
# Q6: Har Embarked port se kitne passengers aaye aur unka average fare kya tha? (groupby + agg — dono ek saath)
# Hint for Q6 — .agg() use hoga:

df.groupby('Embarked').agg({'PassengerId': 'count', 'Fare': 'mean'})
```
